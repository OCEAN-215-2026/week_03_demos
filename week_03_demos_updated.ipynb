## Range objects and for loops

````python
range?
````

We use range often in for loops because it is immutable, simple, and doesn't take up RAM

```python
# BAD: out of memory

out = []
for i in range(1_000_000_000):
    out.append(i)
```

```python
# Works (but wait a long time)

out = 0
for i in range(1_000_000_000):
    out = out + i
```

````python
for i in range(5): # Starts at 0, goes UP TO BUT DOES NOT INCLUDE 5
    print(i)
````

````python
for i in range(3, 10): # Starts at 3, goes UP TO BUT DOES NOT INCLUDE 10
    print(i)
````

````python
for i in range(0, 10, 2): # Starts at 0, goes UP TO 10 BUT DOES NOT INCLUDE 10 by steps of 2
    print(i)
````

````python
for i in range(10, -10, -2): # Starts at 0, goes UP TO 10 BUT DOES NOT INCLUDE 10 by steps of 2
    print(i)
````

````python
for i in range(10, -10.1, -2): # Starts at 0, goes UP TO 10 BUT DOES NOT INCLUDE 10 by steps of 2
    print(i)
````

````python
# convert a range object into a list

list(range(0, 10, 2))
````

````python
# convert a range object into a list

x_list = list(range(0, 10, 2))
x_list.extend([-1, 1])

print(x_list)
````

## Enumerate example

````python
# Example inputs

a_list = [5, 6, 7, 6, 5]
target = 6

# find ALL indices for which the value is matched

indices = []

for i, x in enumerate(a_list):
    if x == target:
        indices.append(i)

print(indices)
````

## Control flow applications

````python
# monthly averages of Seattle water temperature (°F)
seattle_temps = [40.0, 40.6, 44.2, 48.4, 54.9, 60.2, 66.2, 66.7, 60.5, 52.0, 44.5, 39.6]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
````

````python
# Goal: calculate yearly average

total = 0
for x in seattle_temps:
    total = total + x

mean = total / len(seattle_temps)
print(mean)
````

````python
# Goal: construct a list of monthly averages in celsius

temps_C = []
for x in seattle_temps:
    y = (x - 32) * 5 / 9
    temps_C.append(y)

print(temps_C)
````

````python
# Finding hot temperatures
hot = []

for temp in seattle_temps:
    if temp > 55:
        hot.append(temp)

print(hot) 
````

````python
# What if we also want the month that corresponds to the hot temperature?
# There are a few ways to do this, but here's one using enumerate

hot = []
hot_months = []

for index, temp in enumerate(seattle_temps):
    if temp > 55:
        hot.append(temp)
        hot_months.append(index + 1) # Why do we use + 1 here?

print(hot)
print(hot_months)
````

````python
# alternative: extract the short names of hot months
hot = []
hot_months = []

for index, temp in enumerate(seattle_temps):
    if temp > 55:
        hot.append(temp)
        hot_months.append(months[index])

print(hot)
print(hot_months)
````

Let's combine the previous steps: **using a single `for` loop**, generate the following:
+ A list of hot temperatures
+ A list of the corresponding hot months
+ A list of cold temperatures
